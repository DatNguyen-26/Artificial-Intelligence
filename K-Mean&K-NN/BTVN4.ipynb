Mô tả ứng dụng ngắn gọn: Hệ thống tự động phân nhóm tập khách hàng cũ của doanh nghiệp thành 2 phân khúc: Tiết kiệm (Lớp 0) và VIP (Lớp 1) bằng K-Means. Sau đó, khi có thành viên mới đăng ký, hệ thống dùng K-NN đối chiếu với tập dữ liệu cũ để dán nhãn phân loại ngay lập tức.

```python
import numpy as np
from scipy.spatial.distance import cdist

if __name__ == "__main__":
    print("=== THỬ NGHIỆM BÀI 4: ỨNG DỤNG PHÂN KHÚC KHÁCH HÀNG ===")

    # Dữ liệu khách hàng cũ: [Thu nhập tháng (triệu), Điểm số chi tiêu (1-10)]
    Database_KhachHang = np.array([[12, 2], [15, 3], [14, 2], [70, 9], [85, 8], [90, 9]])

    print("Bước 1: Chạy K-Means tự động gắn nhãn phân khúc khách hàng...")
    centers = Database_KhachHang[np.random.choice(Database_KhachHang.shape[0], 2, replace=False)]
    for _ in range(10):
        D = cdist(Database_KhachHang, centers)
        labels_khachhang = np.argmin(D, axis=1)
        new_centers = np.zeros((2, Database_KhachHang.shape[1]))
        for k in range(2):
            Xk = Database_KhachHang[labels_khachhang == k, :]
            if len(Xk) > 0: new_centers[k, :] = np.mean(Xk, axis=0)
        centers = new_centers
    print("   Nhãn phân cụm khách hàng cũ tạo ra tự động:", labels_khachhang)

    print("\nBước 2: Phân loại phân nhóm cho 2 khách hàng mới tham gia hệ thống...")
    KhachHang_Moi = np.array([[13, 2], [75, 8]])

    # Tính KNN thủ công cho Khách hàng mới
    num_test, num_train = KhachHang_Moi.shape[0], Database_KhachHang.shape[0]
    y_pred = np.zeros((num_test, num_train))
    for i in range(num_test):
        for j in range(num_train):
            y_pred[i, j] = np.sqrt(np.sum(np.power(KhachHang_Moi[i, :] - Database_KhachHang[j, :], 2)))

    labels_moi = []
    for i in range(len(y_pred)):
        zipped = zip(y_pred[i, :], labels_khachhang)
        res = sorted(zipped, key=lambda x: x[0])[:3] # k=3
        classes = {}
        for _, j in res:
            j = int(j)
            classes[j] = classes.get(j, 0) + 1
        labels_moi.append(max(classes, key=classes.get))

    for idx, label in enumerate(labels_moi):
        print(f"   Khách hàng mới đặc trưng {KhachHang_Moi[idx]} thuộc về phân nhóm: {label}")
    print("-" * 50)
```
