```python
# ===== ĐỌC FILE MA TRẬN KỀ =====
def read_graph_from_file(filename):
    G = []
    with open(filename, "r") as f:
        for line in f:
            row = list(map(int, line.strip().split()))
            G.append(row)
    return G


# ===== TÔ MÀU ĐỒ THỊ (GREEDY) =====
def graph_coloring(G):
    n = len(G)

    # Tạo tên đỉnh: A, B, C,...
    node = [chr(65 + i) for i in range(n)]

    # Map tên -> index
    t_ = {node[i]: i for i in range(n)}

    # Tính bậc
    degree = [sum(G[i]) for i in range(n)]

    # Màu ban đầu (có thể tăng thêm nếu cần)
    base_colors = ["Blue", "Red", "Yellow", "Green", "Orange", "Purple"]
    colorDict = {node[i]: base_colors.copy() for i in range(n)}

    # ===== SẮP XẾP ĐỈNH THEO BẬC GIẢM =====
    sortedNode = []
    used = []

    for _ in range(n):
        max_deg = -1
        idx = -1
        for j in range(n):
            if j not in used and degree[j] > max_deg:
                max_deg = degree[j]
                idx = j
        used.append(idx)
        sortedNode.append(node[idx])

    # ===== TÔ MÀU =====
    solution = {}

    for u in sortedNode:
        available_colors = colorDict[u]

        # Nếu hết màu → thêm màu mới
        if not available_colors:
            new_color = f"Color{len(base_colors)+1}"
            base_colors.append(new_color)
            available_colors.append(new_color)

        chosen_color = available_colors[0]
        solution[u] = chosen_color

        # Loại màu khỏi đỉnh kề
        for v_idx, is_adj in enumerate(G[t_[u]]):
            if is_adj == 1:
                v = node[v_idx]
                if chosen_color in colorDict[v]:
                    colorDict[v].remove(chosen_color)

    return solution


# ===== MAIN =====
if __name__ == "__main__":
    filename = "graph.txt"   # đổi tên file tại đây
    G = read_graph_from_file(filename)

    result = graph_coloring(G)

    print("Kết quả tô màu:")
    for node, color in sorted(result.items()):
        print(f"Đỉnh {node} = {color}")
```
