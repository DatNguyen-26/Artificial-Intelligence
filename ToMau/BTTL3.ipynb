```python
INF = 10**9

# ===== TSP - NHÁNH CẬN =====
def TSP(cost):
    n = len(cost)
    visited = [False] * n
    path = [0] * n

    best_cost = INF
    best_path = []

    # Cận dưới đơn giản: cạnh nhỏ nhất
    min_edge = min(cost[i][j] for i in range(n) for j in range(n) if i != j)

    def backtrack(i, current_cost):
        nonlocal best_cost, best_path

        # Cắt nhánh
        if current_cost + (n - i) * min_edge >= best_cost:
            return

        # Nếu đã đi đủ n thành phố
        if i == n:
            total_cost = current_cost + cost[path[i-1]][path[0]]
            if total_cost < best_cost:
                best_cost = total_cost
                best_path = path[:]
            return

        # Thử các thành phố tiếp theo
        for j in range(1, n):
            if not visited[j]:
                visited[j] = True
                path[i] = j

                backtrack(i + 1, current_cost + cost[path[i-1]][j])

                visited[j] = False

    # Bắt đầu từ thành phố 0
    visited[0] = True
    path[0] = 0

    backtrack(1, 0)

    return best_cost, best_path


# ===== IN KẾT QUẢ =====
def print_result(cost, path, best_cost):
    node = [chr(65 + i) for i in range(len(cost))]

    print("\n========== KET QUA BAI TOAN TSP ==========\n")
    print(f"So thanh pho: {len(cost)}\n")

    # ===== IN CHU TRINH =====
    print("Chu trinh toi uu:")
    route = " -> ".join(node[i] for i in path) + " -> " + node[path[0]]
    print(route)

    # ===== IN BANG CHI PHI =====
    print("\nBang chi phi tung buoc:")
    print("+---------+---------+---------+")
    print("| Tu      | Den     | Chi phi |")
    print("+---------+---------+---------+")

    total = 0
    for i in range(len(path)):
        u = path[i]
        v = path[(i + 1) % len(path)]
        c = cost[u][v]
        total += c
        print(f"| {node[u]:<7} | {node[v]:<7} | {c:<7} |")

    print("+---------+---------+---------+")

    print(f"\nTong chi phi: {total}")


# ===== MAIN =====
def main():
    # Ví dụ ma trận chi phí
    cost = [
        [0, 10, 15, 20],
        [10, 0, 35, 25],
        [15, 35, 0, 30],
        [20, 25, 30, 0]
    ]

    best_cost, best_path = TSP(cost)
    print_result(cost, best_path, best_cost)


if __name__ == "__main__":
    main()
```
