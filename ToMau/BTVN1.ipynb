```python
import random

# ===== THAM SO =====
DAYS = 5
SLOTS_PER_DAY = 5
TOTAL_SLOTS = DAYS * SLOTS_PER_DAY

teachers = {
    "T1": "Math",
    "T2": "Math",
    "I1": "IT",
    "I2": "IT"
}

teacher_list = list(teachers.keys())
subjects = ["Math", "IT"]

POP_SIZE = 50
GENERATIONS = 200
MUTATION_RATE = 0.1

# ===== TAO CA THE =====
def create_individual():
    individual = []
    for _ in range(TOTAL_SLOTS):
        t = random.choice(teacher_list)
        s = teachers[t]  # môn đúng chuyên môn
        individual.append((t, s))
    return individual


# ===== FITNESS =====
def fitness(ind):
    penalty = 0

    # kiểm tra số tiết mỗi ngày của từng GV
    for d in range(DAYS):
        count = {}
        for s in range(SLOTS_PER_DAY):
            t, _ = ind[d*SLOTS_PER_DAY + s]
            count[t] = count.get(t, 0) + 1

        for t in count:
            if count[t] > 4:
                penalty += (count[t] - 4) * 5

    # cân bằng môn (Math ~ IT)
    math_count = sum(1 for _, s in ind if s == "Math")
    it_count = TOTAL_SLOTS - math_count
    penalty += abs(math_count - it_count)

    return -penalty  # maximize


# ===== SELECTION =====
def selection(pop):
    k = 3
    selected = random.sample(pop, k)
    selected.sort(key=lambda x: fitness(x), reverse=True)
    return selected[0]


# ===== CROSSOVER =====
def crossover(p1, p2):
    point = random.randint(1, TOTAL_SLOTS - 2)
    return p1[:point] + p2[point:]


# ===== MUTATION =====
def mutate(ind):
    if random.random() < MUTATION_RATE:
        idx = random.randint(0, TOTAL_SLOTS - 1)
        t = random.choice(teacher_list)
        ind[idx] = (t, teachers[t])
    return ind


# ===== GA MAIN =====
def genetic_algorithm():
    population = [create_individual() for _ in range(POP_SIZE)]

    for _ in range(GENERATIONS):
        new_pop = []
        for _ in range(POP_SIZE):
            p1 = selection(population)
            p2 = selection(population)
            child = crossover(p1, p2)
            child = mutate(child)
            new_pop.append(child)
        population = new_pop

    best = max(population, key=lambda x: fitness(x))
    return best


# ===== IN KET QUA =====
def print_schedule(ind):
    days_name = ["Thứ Hai", "Thứ Ba", "Thứ Tư", "Thứ Năm", "Thứ Sáu"]

    print("\n========== THỜI KHÓA BIỂU TUẦN ==========\n")

    idx = 0
    count_math = 0
    count_it = 0

    for d in range(DAYS):
        print(f"{days_name[d]}:")
        print("+------+-----------+----------+")
        print("| Tiết | Giáo viên | Môn học  |")
        print("+------+-----------+----------+")

        for s in range(SLOTS_PER_DAY):
            t, sub = ind[idx]

            sub_vn = "Toán" if sub == "Math" else "Tin"

            print(f"| {s+1:<4} | {t:<9} | {sub_vn:<8} |")

            if sub == "Math":
                count_math += 1
            else:
                count_it += 1

            idx += 1

        print("+------+-----------+----------+\n")

    # ===== THỐNG KÊ =====
    print("========== THỐNG KÊ ==========")
    print(f"Tổng số tiết Toán: {count_math}")
    print(f"Tổng số tiết Tin : {count_it}")


# ===== MAIN =====
if __name__ == "__main__":
    best = genetic_algorithm()
    print_schedule(best)
```
