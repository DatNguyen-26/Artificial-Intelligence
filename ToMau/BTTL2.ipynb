```python
# ===== 1. ĐỌC FILE =====
def read_graph(filename):
    G = []
    with open(filename, "r") as f:
        for line in f:
            G.append(list(map(int, line.strip().split())))
    return G


# ===== 2. TẠO DANH SÁCH ĐỈNH =====
def create_nodes(n):
    return [chr(65 + i) for i in range(n)]


# ===== 3. TÍNH BẬC =====
def compute_degree(G):
    return [sum(row) for row in G]


# ===== 4. SẮP XẾP ĐỈNH THEO BẬC GIẢM =====
def sort_nodes_by_degree(nodes, degree):
    sorted_nodes = []
    used = []

    for _ in range(len(nodes)):
        max_deg = -1
        idx = -1
        for i in range(len(nodes)):
            if i not in used and degree[i] > max_deg:
                max_deg = degree[i]
                idx = i
        used.append(idx)
        sorted_nodes.append(nodes[idx])

    return sorted_nodes


# ===== 5. KHỞI TẠO MÀU =====
def init_colors(nodes):
    base_colors = ["Blue", "Red", "Yellow", "Green", "Orange", "Purple"]
    color_dict = {node: base_colors.copy() for node in nodes}
    return color_dict, base_colors


# ===== 6. TÔ MÀU ĐỒ THỊ =====
def color_graph(G, nodes, sorted_nodes, color_dict, base_colors):
    t_ = {nodes[i]: i for i in range(len(nodes))}
    solution = {}

    for u in sorted_nodes:
        available_colors = color_dict[u]

        # Nếu hết màu → thêm màu mới
        if not available_colors:
            new_color = f"Color{len(base_colors)+1}"
            base_colors.append(new_color)
            available_colors.append(new_color)

        chosen_color = available_colors[0]
        solution[u] = chosen_color

        # Loại màu khỏi đỉnh kề
        for v_idx, is_adj in enumerate(G[t_[u]]):
            if is_adj == 1:
                v = nodes[v_idx]
                if chosen_color in color_dict[v]:
                    color_dict[v].remove(chosen_color)

    return solution


# ===== 7. IN KẾT QUẢ =====
def print_result(solution):
    print("Kết quả tô màu:")
    for node, color in sorted(solution.items()):
        print(f"Đỉnh {node} = {color}")


# ===== MAIN =====
def main():
    filename = "graph.txt"
    G = read_graph(filename)

    nodes = create_nodes(len(G))
    degree = compute_degree(G)
    sorted_nodes = sort_nodes_by_degree(nodes, degree)
    color_dict, base_colors = init_colors(nodes)

    solution = color_graph(G, nodes, sorted_nodes, color_dict, base_colors)
    print_result(solution)


if __name__ == "__main__":
    main()
```
